## 3. System curve and operating points

```python
# system curve H = static_head + k * Q^2, calibrated so that 50 L/s needs 28 m
static_head = FILL_IN   # metres of static head
k = FILL_IN             # friction constant from the duty point
system = np.poly1d([k, 0, static_head])
k
```

```python
rho, g, hours, tariff, npsh_avail = FILL_IN   # kg/m3, m/s2, running hours per year, $ per kWh, NPSH available in m
```

```python
rows = []
for p, f in fits.items():
    q = op[p]
    h = f["head"](q)
    eta = f["eff"](q) / 100
    q_bep = FILL_IN    # best-efficiency flow = vertex of the efficiency parabola, -b / (2a)
    margin = FILL_IN   # NPSH available minus NPSHr at the operating flow
    kw = FILL_IN       # shaft power: rho * g * Q(m3/s) * H / eta, in kW
    cost = FILL_IN     # kW * hours * tariff
    rows.append(dict(pump=p, flow_Ls=q, head_m=h, eff_pct=eta * 100, pct_of_bep=100 * q / q_bep,
                     npsh_margin_m=margin, shaft_kw=kw, annual_cost=cost))
res = pd.DataFrame(rows).round(2)
res
```
